# S3 con boto3 y pandas

Cambia el nombre del bucket y la región. Requiere `pip install boto3 pandas s3fs`.

```python
import boto3
import pandas as pd

REGION = "eu-west-3"
BUCKET = "miax-tu-alias-datos"

s3 = boto3.client("s3", region_name=REGION)
```

## Crear un bucket

```python
s3.create_bucket(
    Bucket=BUCKET,
    CreateBucketConfiguration={"LocationConstraint": REGION},
)
```

## Listar buckets y objetos

```python
for bucket in s3.list_buckets()["Buckets"]:
    print(bucket["Name"])
```

```python
for objeto in s3.list_objects_v2(Bucket=BUCKET).get("Contents", []):
    print(objeto["Key"], objeto["Size"])
```

## Subir y descargar con boto3

```python
s3.upload_file("market_data.txt", BUCKET, "market_data.txt")
s3.download_file(BUCKET, "market_data.txt", "copia.txt")
```

## Leer y escribir con pandas

```python
df = pd.read_csv(f"s3://{BUCKET}/market_data.txt", sep=";")
df = df.loc[:, ["FECHA", "HORA", "VALOR", "PRECIO", "VOLUMEN"]]
df.head()
```

```python
df.to_csv(f"s3://{BUCKET}/market_data_mod.csv", index=False)
```
